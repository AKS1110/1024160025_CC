```python
#ANS1
roll_no=input("Enter you roll no.=")
digits=[int(d) for d in roll_no]
l=[d*10 for d in digits]
print("Initial List:",l)
l.append(22)
#22 is added to the list.
print("List after append:",l)
l.insert(4,69)
print("List after insert:",l)
#69 is added to the list at index 4
l.remove(0)
print("List after remove",l)
l.pop(3)
print("List after pop:",l)
l.sort()
print("List after sort:",l)
l.sort(reverse=True)
print("List after sort:",l)
print("First three eles:",l[:3])
print("Last three eles:",l[-3:])
avg=sum(l)/len(l)
print("Avg.=",avg)
greater_than_avg=[c for c in l if c>avg]
print("Eles greater than avg:",greater_than_avg)
```

```python
#ANS2
scores=tuple([d*10 for d in l[:8]])
print("Highest score:",max(scores),",having index:",scores.index(max(scores)))
print("Lowest score:",min(scores),",times it occured:",scores.count(min(scores)))
#Tuples cannot be reversed as they cannot be appended whereas list can can be appended.
reversed_scores=scores[::-1]
print("Reversed scores:",reversed_scores)
eg_score=input("Input a score:")
if eg_score in scores:
  print("First occurrence index:",scores.index(eg_score))
else:
  print("It is not present.")

#This happens because tuple in not mutable.
try:
  scores[0]=100
except TypeError:
  print("Tuple is immutable.")

if len(scores) >= 2:
  first,second,*remaining=scores
  print("First:",first)
  print("second:",second)
  print("Remaining:",remaining)
else:
  print("tuple is empty")





```

```python
#ANS3
import random
your_roll_number=input("Enter your rollno.:")
random.seed(your_roll_number)

numbers=[]
for i in range(100):
  numbers.append(random.randint(100,900))

print("numbers:")
print(numbers)

odd=[]
for x in numbers:
  if x % 2 != 0:
    odd.append(x)

print("Odd Numbers:",odd,",no. of odd nos.:",len(odd))


even=[]
for x in numbers:
  if x % 2 == 0:
    even.append(x)

print("Even Numbers:",even,",no. of even nos.:",len(even))


prime=[]
for x in numbers:
  if x < 2 :
    continue
  is_prime=True

  for i in range(2,x):
    if x % i==0:
       is_prime = False
       break
    if is_prime:
      prime.append(x)

print("Prime Nos.",prime,",No. of prime nos.:",len(prime))

frq_occur=numbers[0]
count=numbers.count(frq_occur)

for x in numbers:
  if numbers.count(x)>count:
    frq_occur=x;
    count=numbers.count(x)
print("Most frequent occuring no.",frq_occur,",No. of times it occured:",count)










```

```python
#ANS4
roll_no=input("Enter your roll no.")
digits=[]
for x in roll_no[:8]:
  digits.append(int(x))

A=set()
B=set()

for x in digits:
  A.add(7*x)
  B.add(9*x)
print("A:",A," ,B:",B)

print("UNION:",A.union(B))

print("INTERSECTION:",A.intersection(B))

print("DIFFERENCE OF A-B",A.difference(B))

print("DIFFERENCE OF B-A",B.difference(A))

print("SYMMETRIC SIFFERENCE:",A.symmetric_difference(B))

print("A is subset of B:",A.issubset(B))

print("B is subset of A:",B.issuperset(A))

x=int(input("Enter a value:"))
A.discard(x)
print("A after discard:",A)

#Discard is safer than remove as it does not show an error when a value is not present.

```
