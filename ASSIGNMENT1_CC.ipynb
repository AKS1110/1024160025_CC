```python
#ANS5.3
def add_primes(n):
    total = 0

    for i in range(2, n + 1):
        prime = True

        for j in range(2, int(i ** 0.5) + 1):
            if i % j == 0:
                prime = False
                break

        if prime:
            total += i

    return total

n = int(input("Enter the value of n: "))
print("Sum =", add_primes(n))
```
