```python
print("Q6 Create a simple dataset :")

employees_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25"
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employees_data)
employees.to_csv("employees.csv", index=False)


print(employees)
```

```python
print("\nQ6(a)  Shape:")

print(employees.shape)

print("\nQ6(b)  DataFrame Information:")

employees.info()

print("\nQ6(c)  Descriptive Statistics:")

print(employees.describe())

print("\nQ6(d)  First 5 rows:")

print(employees.head(5))

print("\nLast 3 rows:")

print(employees.tail(3))

average_salary = employees["Salary"].mean()

print("\nQ6(e)(i) Average Salary:")

print(average_salary)

total_bonus = employees["Bonus"].sum()

print("\nQ6(e)(ii) Total Bonus:")

print(total_bonus)

youngest_age = employees["Age"].min()

print("\nQ6(e)(iii) Youngest Employee's Age:")

print(youngest_age)


highest_rating = employees["Rating"].max()

print("\nQ6(e)(iv) Highest Rating:")

print(highest_rating)

sorted_employees = employees.sort_values(
    by="Salary",
    ascending=False
)

print("\nQ6(f)  Employees sorted by Salary (Descending):")

print(sorted_employees)

def performance_category(rating):

    if rating >= 4.5:
        return "Excellent"

    elif rating >= 4.0:
        return "Good"

    else:
        return "Average"


employees["Performance"] = employees["Rating"].apply(
    performance_category
)

print("\nQ6(g) Performance Category Added:")

print(employees[["Name", "Rating", "Performance"]])


print("\nQ6(h) Missing Values:")

print(employees.isnull().sum())

employees.rename(
    columns={"Employee_ID": "ID"},
    inplace=True
)

print("\nQ6(i)  After Renaming Employee_ID to ID:")

print(employees)

more_than_5 = employees[
    employees["Years_of_Experience"] > 5
]

print("\nQ6(j)(i)  Employees with more than 5 years experience:")

print(more_than_5)

IT_employees = employees[
    employees["Department"] == "IT"
]

print("\nQ6(j)(ii) Employees in IT Department:")

print(IT_employees)

employees["Tax"] = employees["Salary"] * 0.10

print("\nQ6(k) Tax Column Added:")

print(employees[["Name", "Salary", "Tax"]])

employees.to_csv(
    "employees_modified.csv",
    index=False
)

print("\nQ6(l) Modified DataFrame saved as employees_modified.csv")












```
