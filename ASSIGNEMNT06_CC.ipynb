```python
#ANS1
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a.
corrected = temperature + 2
print("Corrected:", corrected)

# b.
fahrenheit = (9/5) * corrected + 32
print("Fahrenheit:", fahrenheit)

# c.
greater = corrected[corrected > 32]
print("Greater than 32:", greater)

# d.
count = np.sum(corrected > 32)
print("Count:", count)
```

```python
#ANS2
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a.
print("Total:", np.sum(steps))

# b.
print("Mean:", np.mean(steps))

# c.
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))

# d.
print("Day totals:", np.sum(steps, axis=0))

# e.
print("User totals:", np.sum(steps, axis=1))

# f.
print("Maximum position:", np.unravel_index(np.argmax(steps), steps.shape))
```

```python
#ANS3
import numpy as np

# a.
original = np.array([1, 2, 3, 4, 5, 6])

# b.
subset = original[1:5]
print("Subset:", subset)

# c.
subset[0] = 999

print("Original:", original)
print("Subset:", subset)

# d.
copied = original[1:5].copy()
copied[0] = 500

print("Original:", original)
print("Copied:", copied)

# e.
matrix = np.arange(1, 13).reshape(3, 4)
print("Matrix:")
print(matrix)

# f.
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2, Columns 2-3:")
print(matrix[1:3, 2:4])

# g.
flat1 = matrix.flatten()
flat2 = matrix.ravel()

print("Flatten:", flat1)
print("Ravel:", flat2)

# h.
flat2[0] = 100
print("After ravel modification:")
print(matrix)

# i.
flat1[1] = 200
print("After flatten modification:")
print(matrix)

# j.
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
#ANS4
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a.
print("Shape:", X.shape)
print("Dimensions:", X.ndim)

# b.
print("Transpose:")
print(X.T)

# c.
XT_X = X.T @ X
print("X.T @ X:")
print(XT_X)

# d.
inverse = np.linalg.inv(XT_X)
print("Inverse:")
print(inverse)

# e.
beta = inverse @ X.T @ y
print("Coefficients:", beta)

# g.
new_user = np.array([5, 40, 7])

prediction = new_user @ beta
print("Predicted assistance score:", prediction)
```
