```python
#ANS2
#(a)
arr = np.array([1,2,3,6,4,5])
rev_arr=arr[::-1]
print("REVERSED ARRAY:",rev_arr)

#(b)
def most_freq(arr):
  vals,counts=np.unique(arr,return_counts=True)
  most_freq_val=vals[np.argmax(counts)]
  indices=np.where(arr==most_freq_val)[0]
  return most_freq_val,indices

x=np.array([1,2,3,4,5,1,2,1,1,1])
val_x,idx_x=most_freq(x)
print(f"x -> Most Frequent:{val_x},Indices:{idx_x}")

y=np.array([1,1,1,2,3,4,2,4,3,3])
val_y,idx_y=most_freq(y)
print(f"y -> Most Frequent: {val_y}, Indices: {idx_y}")

```

```python
#ANS4
amanjot=np.linspace(10,100,25)
print("DIMENSIONS:",amanjot.ndim)
print("SHAPE:",amanjot.shape)
print("TOTAL ELES:",amanjot.size)
print("DATA TYPE:",amanjot.dtype)
print("BYTES CONSUMED:",amanjot.nbytes)

transposed_reshape = amanjot.reshape(25, 1)
print("Transposed via reshape shape:",transposed_reshape.shape)

transposed_T = amanjot.reshape(1, 25).T
print("Transposed via .T shape:",transposed_T.shape)


```

```python
#ANS5
ucs420_amanjot = np.array(
    [[10, 20, 30, 40],[50, 60, 70, 80],[90, 15, 20, 35]]
)

print("MEAN:", np.mean(ucs420_amanjot))
print("MEDIAN:", np.median(ucs420_amanjot))
print("MAX:", np.max(ucs420_amanjot))
print("MIN:", np.min(ucs420_amanjot))
print("UNIQUE ELEMENTS:", np.unique(ucs420_amanjot))

reshaped_ucs420_amanjot = ucs420_amanjot.reshape(4, 3)
print("\nRESHAPED ARRAY(4X3):\n", reshaped_ucs420_amanjot)

# Resize to 2 rows and 3 columns
resized_ucs420_amanjot= np.resize(ucs420_amanjot,(2, 3))
print("\nRESIZED ARRAY(2x3):\n", resized_ucs420_amanjot)
```
