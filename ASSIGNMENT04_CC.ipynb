```python
import pandas as pd

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

roll_number = "1024160025"
last_two_digits = roll_number[-2:]
print("Last two digits:", last_two_digits)
categories = ["billing", "account", "general"]
personalized_entries = []

for digit in last_two_digits:
    d = int(digit)
    category = categories[d % 3]
    if category == "general":
        question = "where can i contact the support team"
        answer = "You can contact the support team through the help desk."
        keywords = "support help contact"
    elif category == "billing":
        question = "how can i check my payment status"
        answer = "You can check your payment status in the billing section."
        keywords = "payment status billing"
    else:
        question = "how do i update my registered mobile number"
        answer = "You can update your mobile number from account settings."
        keywords = "mobile number update"
    personalized_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })
all_entries = fixed_entries + personalized_entries
df = pd.DataFrame(all_entries)

print("Final 6 row datarame")
print(df)
```

```python
def score_query(query, df):
    query_words = query.lower().split()
    results = []
    for index, row in df.iterrows():
        keywords = row["keywords"].lower().split()
        score = 0
        for word in query_words:
            if word in keywords:
                score += 1
        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })
    results = sorted(
        results,
        key=lambda x:x["score"],
        reverse=True
    )

    return results


query = input("\nQ2:Enter your question: ")

results = score_query(query, df)

print("\nMatching FAQs:")

if len(results) == 0:
    print("No matching FAQ found.")

else:
    for result in results:
        print(
            "Score:", result["score"],
            "| Category:", result["category"],
            "| Question:", result["question"]
        )

```

```python
def same_category(category_name, df):

    result = df[df["category"] == category_name]

    return result["question"]


personalized_category = personalized_entries[0]["category"]

print("\nQ3: Questions in category:", personalized_category)

print(
    same_category(
        personalized_category,
        df
    )
)
```

```python
entry_index = 0

print("\nQ4:Selected FAQ:")
print(df.loc[entry_index,"question"])


new_keyword = input("Enter a new keyword: ")


df.loc[entry_index, "keywords"] = (
    df.loc[entry_index, "keywords"]
    + " "
    + new_keyword
)


filename = roll_number + "_faq_data.csv"

df.to_csv(filename, index=False)

print("\nUpdated FAQ:")
print(df.loc[entry_index])

print("\nCSV file saved as:", filename)

```

```python
def score_query_with_ties(query, df):

    results = score_query(query, df)

    if len(results) == 0:
        print("No matching FAQ found.")
        return

    highest_score = results[0]["score"]

    top_results = []

    for result in results:

        if result["score"] == highest_score:
            top_results.append(result)

    print("\nHighest score:", highest_score)

    print("Matching entries:")

    for result in top_results:

        print(
            "Question:",
            result["question"]
        )

        print(
            "Answer:",
            result["answer"]
        )

        print(
            "Category:",
            result["category"]
        )

        print(
            "Score:",
            result["score"]
        )

        print()



print("\nQ6-TIE EXAMPLE")

score_query_with_ties("fee", df)


print("\nQ6-NO TIE EXAMPLE")

score_query_with_ties("password", df)
```
